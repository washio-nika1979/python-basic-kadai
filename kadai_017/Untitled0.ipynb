```python
class Human:
  def __init__(self, name ,age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f'{self.name}は大人です。')
    else:
      print(f'{self.name}は大人ではありません。')

mike = Human('Mike',46)
john = Human('John',35)
dick = Human('Dick',18)
nancy = Human('Nancy',22)
meary = Human('Meary',14)

studens = [mike,john,dick,nancy,meary]

for student in studens:
  student.check_adult()
```
